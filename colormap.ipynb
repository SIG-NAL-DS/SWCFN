```python
import pandas as pd
```

```python


# Define integer class values from the QML
integer_classes = [91, 93, 98, 99, 103, 105, 123, 124, 144, 149, 162, 163, 181, 182, 183, 185, 186, 189, 202]

# Create a DataFrame
rows = []
for i in range(0, 255):  # Assuming raster values range from 0-255
    if i in integer_classes:
        rows.append([i, i, i])  # Keep valid classes
    else:
        rows.append([i, i, -9999])  # Set NoData for decimals

# Save to CSV
df = pd.DataFrame(rows, columns=["From", "To", "NewValue"])

```
